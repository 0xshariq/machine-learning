```python
X = df.drop(['price'],axis=1)
y = df['price']
```

```python
X_one_hot_encode = pd.get_dummies(X,columns=['model','transmission','fuelType'],drop_first=True).astype(int)
X_one_hot_encode.head()
```

```python
encode = LabelEncoder()
columns = ['model','transmission','fuelType']
X_label_encode = X.copy()
for i in columns:
    X_label_encode[i] = encode.fit_transform(X_label_encode[i])
X_label_encode.head()
```

```python
numerical_cols = ['year','mileage','tax','mpg','engineSize']
scaler = StandardScaler()
X_one_hot_encode[numerical_cols] = scaler.fit_transform(X_one_hot_encode[numerical_cols])
X_one_hot_encode
```

```python
X_label_encode = scaler.fit_transform(X_label_encode)
X_label_encode
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_hot_encode,y,test_size=20,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test,y_pred)
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
r2
```

```python
adjusted_r2
```
